```python
import pandas as pd
import numpy as np
```

```python
order_files = ['data/orders_2024_01.csv',
               'data/orders_2024_02.csv',
               'data/orders_2024_03.csv']
orders = pd.concat([pd.read_csv(f) for f in order_files], ignore_index=True)
orders['order_datetime'] = pd.to_datetime(orders['order_datetime'])

print("Заказы:", orders.shape)
print(orders.head())

events_lines = pd.read_json('data/events_lines.json', lines=True)

events_records = pd.read_json('data/events_records.json')

events = (pd.concat([events_lines, events_records], ignore_index=True)
          .drop_duplicates(subset=['event_id']))
events['event_time'] = pd.to_datetime(events['event_time'])

print("\n lines:", events_lines.shape)
print("records:", events_records.shape)
print("уникальные", events.shape)
```

```python
users_with_events = set(events['user_id'].unique())
print("Пользователей с событиями:", len(users_with_events))
print(sorted(users_with_events))
```

```python
users_with_orders = set(orders['user_id'].unique())
print("Пользователей с заказами:", len(users_with_orders))
print(sorted(users_with_orders))
```

```python
all_users = sorted(users_with_events | users_with_orders)
print("Всего уникальных пользователей:", len(all_users))

users = pd.DataFrame({'user_id': all_users})
users['has_events'] = users['user_id'].isin(users_with_events).astype(int)
users['has_orders'] = users['user_id'].isin(users_with_orders).astype(int)

print("\nТаблица пользователей:")
print(users.to_string(index=False))
```

```python
cross = pd.crosstab(users['has_events'], users['has_orders'],
                    rownames=['has_events'], colnames=['has_orders'],
                    margins=True, margins_name='Всего')
print("Перекрёстная таблица:")
print(cross)
```
